# Working with Null values

```python
# Step 1
df.isnull().sum()
```

Replacing the "normalized-losses", "bore" , "stroke" , "horsepower" ,"peak-rpm" with mean 
but "num-of-doors"  with mode as it is a categorical value and most of the cars in general have 4 doors

```python
# First changing thier data types to right format
avg_1 = df["normalized-losses"].astype("float64").mean(axis=0)
print(avg_1)

df["normalized-losses"].replace(np.nan,avg_1,inplace=True)
```

```python
avg_2 = df["bore"].astype("float64").mean(axis=0)
print(avg_2)
```

```python
df["bore"].replace(np.nan, avg_2, inplace = True)
```

```python
avg_3 = df["stroke"].astype("float64").mean(axis=0)
print(avg_3)
```

```python
df["stroke"].replace(np.nan,avg_3, inplace =True)
```

```python
avg_4 = df["horsepower"].astype("float64").mean(axis=0)
print(avg_4)
```

```python
df["horsepower"].replace(np.nan,avg_4, inplace = True)
```

```python
avg_5 = df["peak-rpm"].astype("float64").mean(axis=0)
print(avg_5)
```

```python
df["peak-rpm"].replace(np.nan, avg_5,inplace = True)
```

```python
# Filling the num of doors with mode also we will drop the rows which dont have the price column as it our target variable and any row not having price is not usefull for us
df["num-of-doors"] = df["num-of-doors"].fillna(df["num-of-doors"].mode()[0])
```

```python
# simply drop whole row with NaN in "price" column as row which dont have the price are useless rows we dont want these rows
df.dropna(subset=["price"],axis=0,inplace=True)

# reset index, because we droped two rows
df.reset_index(drop=True,inplace=True)
```

Check for null values now

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
df.head()
```

```python
df[["bore", "stroke"]] = df[["bore", "stroke"]].astype("float64")
df[["normalized-losses"]] = df[["normalized-losses"]].astype("int64")
df[["price"]] = df[["price"]].astype("float64")
df[["peak-rpm"]] = df[["peak-rpm"]].astype("float64")
```

```python
df["horsepower"] = df["horsepower"].astype("int64")
```

df.dtypes

```python
df["city-kmpl"] = round(df["city-mpg"] * 0.425143707,1)
df["highway-kmpl"] = round(df["highway-mpg"] * 0.425143707,1)
```

# Model Development

List of the Regression models we will use:
1) Linear Regression
2) Gradient boosting
3) Random Forest
4) Decision Tree

```python
from sklearn.model_selection import train_test_split
x_train,x_test,Y_train,Y_test = train_test_split(df_new,y,test_size=0.2,random_state = 42 )
```

```python
x_train.shape
```

```python
Y_train.shape
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score,root_mean_squared_error
```

```python
# Object of the Linear Regression model
lr=LinearRegression()
lr.fit(x_train,Y_train)
```

```python
# Storing the predicted value 
y_pred_lr = lr.predict(x_test)
```

```python
print("Linear Regression Results:\n")
print("Training:-",lr.score(x_train,Y_train))
print("R2 results are",r2_score(Y_test,y_pred_lr))
print("MSE :",mean_squared_error(Y_test,y_pred_lr))
print("RMSE :",root_mean_squared_error(Y_test,y_pred_lr))
print("MAE :",mean_absolute_error(Y_test,y_pred_lr))
```

```python
# Object of Gradient Boosting
gb = GradientBoostingRegressor()
gb.fit(x_train,Y_train)
```

```python
y_pred_gb = gb.predict(x_test)
```

```python
print("Gradient Boosting Results:\n")
print("Training:-",gb.score(x_train,Y_train))
print("R2 results are",r2_score(Y_test,y_pred_gb))
print("MSE :",mean_squared_error(Y_test,y_pred_gb))
print("RMSE :",root_mean_squared_error(Y_test,y_pred_gb))
print("MAE :",mean_absolute_error(Y_test,y_pred_gb))
```

```python
# Object of Random Forest
rf = RandomForestRegressor()
rf.fit(x_train,Y_train)
```

```python
y_pred_rf = rf.predict(x_test)
```

```python
print("Random Forest Results:\n")
print("Training:-",rf.score(x_train,Y_train))
print("R2 results are",r2_score(Y_test,y_pred_rf))
print("MSE :",mean_squared_error(Y_test,y_pred_rf))
print("RMSE :",root_mean_squared_error(Y_test,y_pred_rf))
print("MAE :",mean_absolute_error(Y_test,y_pred_rf))
```

```python
# Decision Tree
dtr = DecisionTreeRegressor()
dtr.fit(x_train,Y_train)
```

```python
y_pred_dtr =  dtr.predict(x_test)
```

```python
print("Decision Tree Results:\n")
print("Training:-",dtr.score(x_train,Y_train))
print("R2 results are",r2_score(Y_test,y_pred_dtr))
print("MSE :",mean_squared_error(Y_test,y_pred_dtr))
print("RMSE :",root_mean_squared_error(Y_test,y_pred_dtr))
print("MAE :",mean_absolute_error(Y_test,y_pred_dtr))
```
